```python
import numpy as np

#Need to set a fixed seed for numpy's random number generator 
#Same starting point for all random numbers generated 
#Should be the same random number produced
np.random.seed(42)

data = np.random.randn(100,3)

#Need to randomly pick 20 row-indices out of the 100 availalbe 
#replace = False will let us choose each index only once - 20 distinct rows 
sample_indices = np.random.choice(100,20, replace = False)
sample = data[sample_indices]

#Need to compute the mean
print("Mean:", sample.mean())

#Need to use a relative path so that it saves the file no matter whose computer runs 
#the code 
output_file_path = "out.txt"
with open(output_file_path, 'w') as f:
    f.write(f"Sample Mean: {sample.mean()}\n")

```
